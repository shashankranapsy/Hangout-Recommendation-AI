```python
# ----------------------------------------------
# FINAL SENTIMENT + HANGOUT RECOMMENDER
# ----------------------------------------------

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# ------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------

df = pd.read_csv(
    "training.1600000.processed.noemoticon.csv",
    encoding="latin-1",
    header=None,
    low_memory=False
)


df.columns = ["label", "id", "date", "query", "user", "text"]


df = df[ df["label"].isin(["0", "4", 0, 4]) ]


df["label"] = df["label"].astype(int)


df = df[df["label"].isin([0, 4])]


df["label"] = df["label"].map({0: 0, 4: 1})


df = df.dropna(subset=["text"])


df_small = df.sample(10000, random_state=42)

print("Final dataset size:", df_small.shape)

X = df_small["text"].astype(str)
y = df_small["label"].astype(int)


model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("clf", LogisticRegression(max_iter=200))
])

model.fit(X, y)

print("Model trained successfully!")


def analyze_chat(text):
    sentiment_score = model.predict_proba([text])[0][1]
    if sentiment_score > 0.6:
        mood = "positive"
    elif sentiment_score < 0.4:
        mood = "negative"
    else:
        mood = "neutral"
    return mood, sentiment_score


def recommend_place(mood):
    positive_places = [
        "CyberHub", "PVR Cinemas", "Bowling Alley",
        "Cafe Coffee Day", "Ice Cream Parlour", "Ambience Mall"
    ]
    negative_places = [
        "Quiet Park", "Library Cafe", "Art Gallery",
        "Lakeside Walk", "Long Drive"
    ]
    neutral_places = [
        "Rooftop Cafe", "Casual Restaurant",
        "Food Court", "Tea Stall"
    ]

    if mood == "positive":
        return ", ".join(positive_places)
    elif mood == "negative":
        return ", ".join(negative_places)
    else:
        return ", ".join(neutral_places)


print("\nENTER CHAT BELOW:\n")

chat = input("Chat:\n")

mood, score = analyze_chat(chat)
print(f"\nMood: {mood.upper()} (score = {score:.2f})\n")

print("Recommended Places:")
print(recommend_place(mood))
```
